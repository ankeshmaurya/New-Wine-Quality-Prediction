```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_predict, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import joblib, json, os
import seaborn as sns
```

```python
df = pd.read_csv("wine.csv")
df.head()
```

```python
X = df.drop("Quality", axis=1)
y = df["Quality"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        class_weight="balanced",
        random_state=42
    ))
])

pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred, labels=sorted(y.unique()))
ConfusionMatrixDisplay(cm, display_labels=sorted(y.unique())).plot(cmap="Blues")
plt.show()
```

```python
plt.figure(figsize=(12,8))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Correlation Heatmap of Wine Features")
plt.show()
```

```python
df.hist(bins=20, figsize=(15,12), edgecolor="black")
plt.suptitle("Feature Distributions", fontsize=16)
plt.show()
```

```python
sns.countplot(x="Quality", data=df, palette="viridis")
plt.title("Distribution of Wine Quality")
plt.show()
```

```python
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
y_pred_oof = cross_val_predict(pipe, X, y, cv=skf, method="predict")
y_pred_proba_oof = cross_val_predict(pipe, X, y, cv=skf, method="predict_proba")

df["predicted_quality_oof"] = y_pred_oof
df["predicted_confidence_oof"] = y_pred_proba_oof.max(axis=1)

# add class probability columns
for i, cls in enumerate(pipe.classes_):
    df[f"proba_{cls}"] = y_pred_proba_oof[:, i]

df.to_csv("wine_with_predictions.csv", index=False)
df.head()
```

```python
joblib.dump(pipe, "wine_model.pkl")

schema = {
    "features": list(X.columns),
    "feature_ranges": {
        col: [float(df[col].min()), float(df[col].max())]
        for col in X.columns
    }
}
json.dump(schema, open("wine_schema.json", "w"))
```
